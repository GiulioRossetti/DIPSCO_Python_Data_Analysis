# Case Study: Songs

```python
#artist = 'gazebo-penguins'
artist = 'radiohead'
url = 'https://lyricstranslate.com/en/'+artist+'-lyrics.html'

url
```

```python
from bs4 import BeautifulSoup
import requests

soup = BeautifulSoup(requests.get(url).text)
```

```python
links = list()
for link in soup.find_all('a'):
    links.append(link.get('href'))
    
links
```

```python
base = 'https://lyricstranslate.com'
for link in links:
    if link and link.lower().startswith('/en/'+artist+'-') and link.endswith('lyrics.html'):
        print(requests.get(base+link).text)
        break
```

```python
import time
base = 'https://lyricstranslate.com'

texts = dict()
for i, link in enumerate(links):
    if link and link.lower().startswith('/en/'+artist+'-') and link.endswith('lyrics.html'):
        print(link)
        
        html = requests.get(base+link).text
        soup = BeautifulSoup(html)
        #texts[soup.find("h2", {"class": "title-h2"}).get_text().strip()] = soup.find("div", {"id": "song-body"}).get_text()
        texts[link.rsplit('/')[2]] = soup.find("div", {"id": "song-body"}).get_text()
        
        time.sleep(1)
        
        #if i >= 50:
        #    break
```

```python
texts.keys()
```

```python
import json

with open(artist+'.json',mode='wt',encoding='utf-8') as outputfile:
    
    json.dump(texts,outputfile)
```
