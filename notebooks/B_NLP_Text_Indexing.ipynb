# Text Indexing with NLTK and SpaCy

```python
### Language Indexing Concepts with NLTK and spaCy

# Install necessary libraries
# !pip install nltk spacy
# !python -m spacy download en_core_web_sm

# Import libraries
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk import pos_tag, ne_chunk, word_tokenize
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.util import ngrams
from collections import Counter

import spacy
from spacy import displacy

# Download NLTK resources
nltk.download('punkt')
nltk.download('averaged_perceptron_tagger')
nltk.download('wordnet')
nltk.download('stopwords')
nltk.download('maxent_ne_chunker')
nltk.download('words')
```

```python
text = """Hunter x Hunter follows Gon Freecss, a boy striving to become a Hunter and find his father.
        Alongside friends Killua, Leorio, and Kurapika, he faces challenges like the Hunter Exam, mastering Nen,
        and battling foes like the Phantom Troupe and Chimera Ants in a story of growth, friendship, and survival."""
```

# NLTK

```python
### NLTK Implementation

# 1. Sentence Splitting with NLTK
print("\n--- Sentence Splitting (NLTK) ---")
print("\n")

sentences = sent_tokenize(text)
print(sentences)
```

```python
text[0]
```

```python
sentences[0]
```

```python
# 2. Tokenization with NLTK
print("\n--- Tokenization (NLTK) ---")
print("\n")

tokens = word_tokenize(text)
print(tokens)
```

```python
from nltk.corpus import stopwords
import string

en_stop = set(stopwords.words('english'))
punctuations = list(set(string.punctuation))
punctuations = set(punctuations)

tokens = [token for token in tokens if token not in en_stop and token not in punctuations]
```

```python
tokens
```

```python
tokens[0]
```

```python
# 3. Part-of-Speech Tagging with NLTK
print("\n--- Part-of-Speech Tagging (NLTK) ---")
print("\n")

pos_tags = pos_tag(tokens)
print(pos_tags)
```

```python
pos_tags[0]
```

```python
# 4. Stemming and Lemmatization with NLTK
print("\n--- Stemming and Lemmatization (NLTK) ---")
print("\n")

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

stems = [stemmer.stem(token) for token in tokens]
lemmas = [lemmatizer.lemmatize(token) for token in tokens]
```

```python
print("Stems:", stems)
```

```python
print("Lemmas:", lemmas)
```

```python
# 5. Bag-of-Words and N-Grams with NLTK
print("\n--- Bag-of-Words and N-Grams (NLTK) ---")
print("\n")

bow = Counter(tokens)
print("Bag-of-Words:", bow)
```

```python
n_grams = list(ngrams(tokens, 2))

print("Bi-grams:", n_grams)
```

```python
# 6. Chunking with NLTK
print("\n--- Chunking (NLTK) ---")
print("\n")

named_entities_tree = ne_chunk(pos_tags)
print(named_entities_tree)
```

```python
# 7. Named Entity Recognition with NLTK
print("\n--- Named Entity Recognition (NLTK) ---")
print("\n")

named_entities = ne_chunk(pos_tags)
named_entities_list = [chunk for chunk in named_entities if hasattr(chunk, 'label')]
print("Named Entities:", named_entities_list)
```

```python
# 8. Parsing (Simplified Example)
print("\n--- Parsing (NLTK) ---")
print("\n")

from nltk.parse import CoreNLPParser
# Ensure you have CoreNLP running locally or replace with a simple example.
print("Parsing can be done using external tools like CoreNLP or syntactic grammars in NLTK.")
```

```python
# 9. Stopwords with NLTK
print("\n--- Stopwords (NLTK) ---")
print("\n")

stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word.lower() not in stop_words]

print("Filtered Tokens:", filtered_tokens)
```

# spaCy

```python
### spaCy Implementation

# Load spaCy model
nlp = spacy.load("en_core_web_sm")
doc = nlp(text)
```

```python
doc
```

```python
doc.sents
```

```python
# 1. Sentence Splitting with spaCy
print("\n--- Sentence Splitting (spaCy) ---")
print("\n")

sentences_spacy = [sent.text for sent in doc.sents]
print(sentences_spacy)
```

```python
# 2. Tokenization with spaCy
print("\n--- Tokenization (spaCy) ---")
print("\n")

tokens_spacy = [token.text for token in doc]
print(tokens_spacy)
```

```python
tokens_spacy = [token for token in tokens_spacy if token not in en_stop and token not in punctuations]

print(tokens_spacy)
```

```python
tokens_spacy = [token.text for token in doc]

tokens_spacy = [token for token in doc if not token.is_stop and not token.is_punct]
```

```python
print(tokens_spacy)
```

```python
# 3. Part-of-Speech Tagging with spaCy
print("\n--- Part-of-Speech Tagging (spaCy) ---")
print("\n")

pos_tags_spacy = [(token.text, token.pos_) for token in doc if not token.is_stop and not token.is_punct]
print(pos_tags_spacy)
```

```python
# 4. Stemming and Lemmatization with spaCy
print("\n--- Lemmatization (spaCy) ---")
print("\n")

lemmas_spacy = [token.lemma_ for token in doc if not token.is_stop and not token.is_punct]
print("Lemmas:", lemmas_spacy)

# Note: Stemming is not natively supported in spaCy.
```

```python
# 5. Bag-of-Words and N-Grams with spaCy
print("\n--- Bag-of-Words and N-Grams (spaCy) ---")
print("\n")

bow_spacy = Counter([token.text for token in doc if not token.is_stop and not token.is_punct])
print("Bag-of-Words:", bow_spacy)
```

```python
n_grams_spacy = [doc[i:i+2].text for i in range(len(doc)-1)]
print("Bi-grams:", n_grams_spacy)
```

```python
# 6. Chunking with spaCy
print("\n--- Chunking (spaCy) ---")
print("\n ")

chunks_spacy = [chunk.text for chunk in doc.noun_chunks]
print("Chunks:", chunks_spacy)
```

```python
# 7. Named Entity Recognition with spaCy
print("\n--- Named Entity Recognition (spaCy) ---")
print("\n")

entities_spacy = [(ent.text, ent.label_) for ent in doc.ents]
print("Entities:", entities_spacy)
```

```python
# 8. Parsing with spaCy
print("\n--- Parsing (spaCy) ---")
for token in doc:
    print(f"{token.text:10} {token.dep_:10} {token.head.text:10} {token.head.pos_:10}")
    
```

```python
# 9. Stopwords with spaCy
print("\n--- Stopwords (spaCy) ---")
print("\n")
stopwords_spacy = spacy.lang.en.stop_words.STOP_WORDS
filtered_tokens_spacy = [token.text for token in doc if token not in stopwords_spacy]
print("Filtered Tokens:", filtered_tokens_spacy)
```
