```python
import json

artist = 'radiohead'
with open(artist + '.json', mode='rt', encoding='utf-8') as inputfile:
    loaded_texts = json.load(inputfile)
    
texts = list(loaded_texts.values())

texts[0]
```

```python
import nltk

nltk.download('punkt')

from nltk import word_tokenize

for text in texts:
    tokens = word_tokenize(text.lower())
    print(tokens)
    break
```

```python
tok_sents = []
for text in texts:
    tokens = word_tokenize(text.lower())
    tok_sents.append(tokens)
```

```python
import nltk
nltk.download('vader_lexicon')
```

```python
from nltk.sentiment.vader import SentimentIntensityAnalyzer

vader = SentimentIntensityAnalyzer()
```

```python
vader.polarity_scores(' '.join(tok_sents[0])) # come back to string
```

```python
from textblob import TextBlob
```

```python
blob = TextBlob(' '.join(tok_sents[0]))
```

```python
def textblob_subjectivity_analysis(sentence):
    blob = TextBlob(sentence)
    return blob.sentiment.subjectivity


subjectivity_scores = [textblob_subjectivity_analysis(text) for text in texts]
subjectivity_scores
```

# BERT for Emotion Detection

```python
from transformers import pipeline
#import torch
#from datasets import load_dataset

# Load a pre-trained emotion detection model from HuggingFace
emotion_model = pipeline("text-classification", model="j-hartmann/emotion-english-distilroberta-base")
```

```python
emotion_model(' '.join(tok_sents[0]))
```

```python
emotion_model(' '.join(tok_sents[1]))
```

```python
for title, text in loaded_texts.items():
    tokens = word_tokenize(text.lower())
    emot = emotion_model(' '.join(tokens))
    
    split_parts = title.split('-')
    split_parts_excluded = split_parts[1:-1]
    name = '-'.join(split_parts_excluded)
    
    print(name, ':', emot[0]['label'], emot[0]['score'])
    print('\n')
```

```python
from transformers import pipeline

# Load the emotion detection model
emotion_model = pipeline("text-classification", model="bhadresh-savani/bert-base-uncased-emotion")
```

```python
for title, text in loaded_texts.items():
    tokens = word_tokenize(text.lower())
    emot = emotion_model(' '.join(tokens))
    
    split_parts = title.split('-')
    split_parts_excluded = split_parts[1:-1]
    name = '-'.join(split_parts_excluded)
    
    print(name, ':', emot[0]['label'], emot[0]['score'])
    print('\n')
```
